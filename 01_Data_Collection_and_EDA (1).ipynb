```python
import zipfile
import os

zip_path = '/content/rossmann-store-sales.zip'
extract_path = '/content/rossmann'

with zipfile.ZipFile(zip_path, 'r') as zip_ref:
    zip_ref.extractall(extract_path)

print("Dataset extracted successfully!")

print(os.listdir(extract_path))
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

print("Libraries imported successfully!")
```

```python
train = pd.read_csv(
    '/content/rossmann/train.csv'
)

store = pd.read_csv(
    '/content/rossmann/store.csv'
)

test = pd.read_csv(
    '/content/rossmann/test.csv'
)

print("Train shape:", train.shape)
print("Store shape:", store.shape)
print("Test shape:", test.shape)

display(train.head())
```

```python
print("TRAIN DATA INFORMATION")
print("=" * 50)

train.info()

print("\nFirst 5 Rows:")
display(train.head())

print("\nStatistical Summary:")
display(train.describe())
```

```python
print("MISSING VALUES IN TRAIN")
print(train.isnull().sum())

print("\nMISSING VALUES IN STORE")
print(store.isnull().sum())

print("\nDUPLICATES")
print("Train duplicates:", train.duplicated().sum())
print("Store duplicates:", store.duplicated().sum())

print("\nDATE RANGE")
print("Start:", train['Date'].min())
print("End:", train['Date'].max())

print("\nUNIQUE STORES")
print(train['Store'].nunique())
```

```python
df = train.merge(
    store,
    on='Store',
    how='left'
)

df['Date'] = pd.to_datetime(df['Date'])

print("Merged Dataset Shape:", df.shape)

display(df.head())
```
